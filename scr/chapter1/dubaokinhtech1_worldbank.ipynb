## Thu Thập Dữ Liệu Kinh Tế Vĩ Mô (GDP, CPI, Dân Số) Của Việt Nam Từ Ngân Hàng Thế Giới

### I. THU THẬP DỮ LIỆU WORLD BANK VỚI R (`WDI` Package)

World Bank cung cấp API mở cho phép truy xuất trực tiếp hàng ngàn chỉ số kinh tế vĩ mô thông qua gói thư viện `WDI` trên R.

**Các mã chỉ số phổ biến:**
- **GDP (Current US$)**: `NY.GDP.MKTP.CD`
- **Dân số (Total Population)**: `SP.POP.TOTL`
- **Chỉ số giá tiêu dùng CPI**: `FP.CPI.TOTL`
- **Mã quốc gia Việt Nam (ISO-2 Code)**: `VN`

#### 1. Tải dữ liệu GDP của Việt Nam
Sử dụng thư viện `WDI` để tải chuỗi thời gian GDP của Việt Nam (`country="VN"`) từ năm 2000 đến 2024.

````r
library(WDI)

# Tai du lieu GDP cua Viet Nam giai doan 2000 - 2024
vn_gdp <- WDI(country = "VN",
              indicator = "NY.GDP.MKTP.CD",
              start = 2000, end = 2024)

# Kiem tra du lieu
head(vn_gdp)
````

#### 2. Tải dữ liệu của cùng lúc nhiều biến
Khai báo vector các chỉ số cần tải `ind` bao gồm GDP (`NY.GDP.MKTP.CD`) và Dân số (`SP.POP.TOTL`) để truy xuất đồng thời.

````r
# Tai du lieu cua cung luc nhieu bien
ind <- c(GDP = "NY.GDP.MKTP.CD", POP = "SP.POP.TOTL")
data <- WDI(country = "VN", indicator = ind)

# Kiem tra du lieu nhieu bien
head(data)
str(data)
````

#### 3. Chuẩn hóa & Trực quan hóa dữ liệu với R
Sử dụng `dplyr` và `ggplot2` để sắp xếp chuỗi thời gian và vẽ biểu đồ tăng trưởng GDP Việt Nam.

````r
library(dplyr)
library(ggplot2)

# Sap xep du lieu theo nam va tinh GDP binh quan dau nguoi
data_clean <- data %>%
  arrange(year) %>%
  mutate(gdp_per_capita = GDP / POP)

# Ve bieu do dien bien GDP Viet Nam
ggplot(data_clean, aes(x = year, y = GDP / 1e9)) +
  geom_line(color = "#0073C2FF", linewidth = 1) +
  geom_point(color = "#0073C2FF", size = 2) +
  labs(title = "Dien bien GDP Viet Nam (World Bank WDI)",
       x = "Nam", y = "GDP (Ty USD)") +
  theme_minimal()
````

### II. THU THẬP DỮ LIỆU WORLD BANK VỚI STATA (`wbopendata`)

Trong phần mềm Stata, module `wbopendata` được sử dụng để tải trực tiếp dữ liệu từ Ngân hàng Thế giới.
- **Mã quốc gia Việt Nam (ISO-3 Code trong Stata)**: `VNM`

#### 1. Tải dữ liệu GDP của Việt Nam trong Stata
Sử dụng câu lệnh `wbopendata` với lựa chọn `country(VNM)` và `indicator(NY.GDP.MKTP.CD)`.

```stata
* Cai dat module neu chua co:
* ssc install wbopendata

* Tai du lieu GDP cua Viet Nam
wbopendata, country(VNM) indicator(NY.GDP.MKTP.CD) clear
```

#### 2. Thu thập GDP, CPI và Dân số Việt Nam cùng lúc trong Stata
Khi thu thập nhiều chỉ số trong Stata, các mã chỉ số được nối với nhau bằng dấu chấm phẩy `;` trong option `indicator()`.

```stata
* Thu thap GDP, CPI va Dan so Viet Nam
wbopendata, country(VNM) ///
indicator( NY.GDP.MKTP.CD;SP.POP.TOTL;FP.CPI.TOTL) clear
```

#### 3. Chuẩn hóa & Thiết lập Chuỗi Thời Gian trong Stata
Sau khi nạp dữ liệu, sắp xếp theo năm và khai báo định dạng chuỗi thời gian với lệnh `tsset`.

```stata
* Xem 10 quan sat dau tien
list countryname year ny_gdp_mktp_cd sp_pop_totl fp_cpi_totl in 1/10

* Sap xep theo thoi gian
sort year

* Thiet lap chuoi thoi gian theo nam
tsset year

* Ve do thi chuoi thoi gian GDP Viet Nam
tsline ny_gdp_mktp_cd, title("Dien bien GDP Viet Nam") xtitle("Nam") ytitle("GDP (USD)")
```
